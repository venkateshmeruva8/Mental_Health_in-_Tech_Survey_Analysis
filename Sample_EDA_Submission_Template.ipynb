## 3. ***Data Wrangling***

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.
df_clean = df.copy()
print("Original shape:", df_clean.shape)
df_clean["Age"] = df_clean["Age"].where(
    df_clean["Age"].between(18, 100),
    np.nan
)

def clean_gender(value):
    value = str(value).strip().lower()

    if value in ["male", "m", "man", "cis male", "male (cis)"]:
        return "Male"
    elif value in ["female", "f", "woman", "cis female", "female (cis)"]:
        return "Female"
    elif value == "nan":
        return np.nan
    else:
        return "Other"

df_clean["Gender_Clean"] = df_clean["Gender"].apply(clean_gender)

bins = [17, 25, 35, 45, 55, 65, 100]
labels = ["18-25", "26-35", "36-45", "46-55", "56-65", "66+"]

df_clean["Age_Group"] = pd.cut(
    df_clean["Age"],
    bins=bins,
    labels=labels
)

categorical_columns = df_clean.select_dtypes(include="object").columns

for column in categorical_columns:
    df_clean[column] = df_clean[column].apply(
        lambda x: x.strip() if isinstance(x, str) else x
    )

print("Cleaned shape:", df_clean.shape)
````

````python
# checking the cleaned dataset
print("Age summary:")
print(df_clean["Age"].describe())

print("\nGender:")
print(df_clean["Gender_Clean"].value_counts(dropna=False))

print("\nAge groups:")
print(df_clean["Age_Group"].value_counts().sort_index())

print("\nDuplicate rows:")
print(df_clean.duplicated().sum())
````

### What all manipulations have you done and insights you found?

Answer Here.
The original dataset was copied into a separate cleaned dataframe named df_clean. Invalid Age values outside the range of 18 to 100 were treated as missing values. The Gender variable was standardized into cleaner categories using a new Gender_Clean variable.

Age_Group was created to divide respondents into meaningful age ranges for further analysis. Whitespace was also removed from categorical variables to improve consistency.

The original dataset was preserved, while df_clean is used for further exploratory analysis.

#### Chart - 14 - Correlation Heatmap

````python
corr_data = df_clean[
    [
        "Age",
        "treatment",
        "family_history",
        "remote_work",
        "tech_company",
        "obs_consequence"
    ]
].copy()

binary_columns = [
    "treatment",
    "family_history",
    "remote_work",
    "tech_company",
    "obs_consequence"
]

for column in binary_columns:
    corr_data[column] = corr_data[column].map({"Yes": 1, "No": 0})

correlation_matrix = corr_data.corr()

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Correlation Heatmap of Selected Variables")
plt.tight_layout()
plt.show()
````

##### 1. Why did you pick the specific chart?

A correlation heatmap is selected to visualize the strength and direction of relationships between selected numerical and binary variables. It provides an easy way to identify variables that have relatively stronger or weaker correlations with each other.

##### 2. What is/are the insight(s) found from the chart?

The heatmap shows that Family History and Treatment have the strongest correlation among the selected variables, with a correlation value of approximately 0.38.

Treatment also has a correlation of approximately 0.16 with Observation Consequence. Age has a small positive correlation with Remote Work of approximately 0.15.

Most of the other variable pairs have correlation values close to zero, indicating relatively weak linear relationships. For example, Age and Family History have a correlation of approximately 0.01, while Remote Work and Observation Consequence have approximately -0.04.

Overall, the heatmap indicates that Family History has the strongest relationship with Treatment among the variables selected for this analysis, while most other relationships are relatively weak.

#### Chart - 15 - Pair Plot

````python
# Pair Plot visualization code
pair_data = df_clean[
    [
        "Age",
        "treatment",
        "family_history",
        "remote_work",
        "tech_company"
    ]
].copy()

for column in ["treatment", "family_history", "remote_work", "tech_company"]:
    pair_data[column] = pair_data[column].map({"Yes": 1, "No": 0})

pair_data = pair_data.dropna()

sns.pairplot(
    pair_data,
    diag_kind="hist"
)

plt.suptitle(
    "Pair Plot of Selected Variables",
    y=1.02
)

plt.show()
````

##### 1. Why did you pick the specific chart?

A pair plot is selected because it allows multiple numerical variables to be compared simultaneously. It shows the distribution of each variable and the relationships between pairs of variables in a single visualization.

##### 2. What is/are the insight(s) found from the chart?

The pair plot shows that Age has a wider numerical distribution, while Treatment, Family History, Remote Work, and Tech Company are binary variables with values of 0 and 1.

The plots involving the binary variables form distinct groups around 0 and 1. The relationship between Age and the binary variables does not show a strong clear linear pattern.

The diagonal plots show the individual distributions of the selected variables, while the off-diagonal plots show their pairwise relationships.

Overall, the pair plot provides a visual confirmation that the selected categorical/binary variables form separate groups and that Age does not show an obvious strong relationship with these variables.
